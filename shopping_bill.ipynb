```python
# Shopping Bill Generator
products = []
# Get the number of products
try:
    n = int(input("Enter the number of products: "))
    if n <= 0:
        print("Please enter a positive number of products.")
    else:
        for i in range(n):
            print("\nProduct", i + 1)
            name = input("Enter product name: ")
            # Validate price
            try:
                price = float(input("Enter unit price: "))
                if price <= 0:
                    print("Price must be greater than zero.")
                    continue
                # Validate quantity
                quantity = int(input("Enter quantity: "))
                if quantity <= 0:
                    print("Quantity must be a positive whole number.")
                    continue
                # Calculate subtotal
                subtotal = price * quantity
                # Store product details
                product = {
                    "name": name,
                    "price": price,
                    "quantity": quantity,
                    "subtotal": subtotal
                }
                products.append(product)
            except ValueError:
                print("Invalid input! Enter a valid price and quantity.")
        # Generate bill
        if len(products) > 0:
            print("\n========== SHOPPING BILL ==========")
            total = 0
            for product in products:
                print("\nProduct Name:", product["name"])
                print("Unit Price: ₹", product["price"])
                print("Quantity:", product["quantity"])
                print("Subtotal: ₹", product["subtotal"])
                total = total + product["subtotal"]
            # Find the most expensive product
            expensive = products[0]
            for product in products:
                if product["price"] > expensive["price"]:
                    expensive = product
            # Calculate discount
            if total >= 2000:
                discount = total * 0.10
            else:
                discount = 0
            # Calculate tax after discount
            amount_after_discount = total - discount
            tax = amount_after_discount * 0.05
            # Final amount
            final_amount = amount_after_discount + tax
            # Print final receipt
            print("\n========== FINAL RECEIPT ==========")
            print("Total Before Discount: ₹", round(total, 2))
            print("Discount Amount: ₹", round(discount, 2))
            print("Tax Amount (5%): ₹", round(tax, 2))
            print("Final Payable Amount: ₹", round(final_amount, 2))
            print("\nMost Expensive Product:", expensive["name"])
            print("Unit Price: ₹", expensive["price"])
            print("\nThank you for shopping!")
        else:
            print("No valid products were added.")
except ValueError:
    print("Invalid input! Enter a whole number for the number of products.")
```
